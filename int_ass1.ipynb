```python
#10 sum of first 20 odd numbers
sum_odd = 0
count = 0
num = 1
while count < 20:
    if num % 2 != 0:
        sum_odd += num
        count += 1
    num += 1
print(sum_odd)
```

```python
n = 20
sum_odd = sum([2*i + 1 for i in range(n)])
print(f"Sum of first {n} odd numbers is {sum_odd}")
```
